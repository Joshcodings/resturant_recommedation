```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor, HistGradientBoostingRegressor

def evaluate(model, X_tr, X_te, y_tr, y_te):
    model.fit(X_tr, y_tr)
    tr_pred = model.predict(X_tr)
    te_pred = model.predict(X_te)
    return {
        "train_R2": r2_score(y_tr, tr_pred),
        "test_MAE": mean_absolute_error(y_te, te_pred),
        "test_R2": r2_score(y_te, te_pred),
    }
```

```python
models = {
    "Linear Regression": LinearRegression(),
    "Random Forest": RandomForestRegressor(n_estimators=300, min_samples_leaf=3,
                                           random_state=42, n_jobs=-1),
    "Gradient Boosting": HistGradientBoostingRegressor(max_iter=300, learning_rate=0.05,
                                                       random_state=42),
}

rows = []
for name, m in models.items():
    r = evaluate(m, X_train, X_test, y_train, y_test)
    rows.append({"model": name, "inputs": "A: no votes", **r})
```

```python
X_B = pd.get_dummies(rated[features_B], columns=cat_cols, dtype=int)
XB_train, XB_test, yB_train, yB_test = train_test_split(
    X_B, y, test_size=0.2, random_state=42
)

for name, m in models.items():
    r = evaluate(m, XB_train, XB_test, yB_train, yB_test)
    rows.append({"model": name, "inputs": "B: with votes", **r})
```

```python
results = pd.DataFrame(rows).round(3)
baseline_row = pd.DataFrame([{"model": "Baseline (mean)", "inputs": "-",
                              "train_R2": 0.0, "test_MAE": 0.46, "test_R2": 0.0}])
pd.concat([baseline_row, results], ignore_index=True)
```

```python
from sklearn.inspection import permutation_importance

gb_A = HistGradientBoostingRegressor(max_iter=300, learning_rate=0.05, random_state=42)
gb_A.fit(X_train, y_train)

r = permutation_importance(gb_A, X_test, y_test, n_repeats=10, random_state=42,
                           scoring="neg_mean_absolute_error", n_jobs=-1)
imp = pd.Series(r.importances_mean, index=X_test.columns).sort_values(ascending=False)
imp.head(15).round(4)
```

```python
pred = gb_A.predict(X_test)
check = rated.loc[X_test.index, ["country"]].copy()
check["region"] = np.where(check["country"] == "India", "India", "Other countries")
check["abs_err"] = (y_test - pred).abs()
check["residual"] = y_test - pred

check.groupby("region").agg(n=("abs_err", "size"),
                            MAE=("abs_err", "mean"),
                            avg_bias=("residual", "mean")).round(3)
```

```python
q80 = np.quantile((y_test - pred).abs(), 0.80)
print("80% of predictions are within ±", round(q80, 2), "rating points")
```

```python
def cols_without(prefixes):
    return [c for c in X.columns if not any(c.startswith(p) for p in prefixes)]

tests = {
    "All features":              list(X.columns),
    "No city_group":             cols_without(["city_group_"]),
    "No country":                cols_without(["country_"]),
    "No city_group, no country": cols_without(["city_group_", "country_"]),
    "No cuisines":               cols_without(["cuisine_"]),
}

rows = []
for name, cols in tests.items():
    m = HistGradientBoostingRegressor(max_iter=300, learning_rate=0.05, random_state=42)
    r = evaluate(m, X_train[cols], X_test[cols], y_train, y_test)
    rows.append({"feature set": name, "n_features": len(cols),
                 "test_MAE": r["test_MAE"], "test_R2": r["test_R2"]})

pd.DataFrame(rows).round(3)
```

## diverting to recommendation system 

```python
reco = df[df["is_rated"] == 1].copy()
reco["cuisine_list"] = reco["cuisines"].str.split(",").apply(lambda xs: [x.strip() for x in xs])
print("Restaurants available to recommend:", len(reco))
print("Unique cuisines:", reco["cuisine_list"].explode().nunique())
```

```python
m = 50
city_mean = reco.groupby("city")["aggregate_rating"].transform("mean")
v, R = reco["votes"], reco["aggregate_rating"]

reco["weighted_rating"] = (v / (v + m)) * R + (m / (v + m)) * city_mean
reco["weighted_rating"] = reco["weighted_rating"].round(3)

cols = ["restaurant_name", "city", "aggregate_rating", "votes", "weighted_rating"]
reco.sort_values("weighted_rating", ascending=False)[cols].head(10)
```

```python
print("Restaurants rated 4.5+ with under 20 votes:",
      ((reco["aggregate_rating"] >= 4.5) & (reco["votes"] < 20)).sum())
print("Restaurants rated 4.5+ with 200+ votes:",
      ((reco["aggregate_rating"] >= 4.5) & (reco["votes"] >= 200)).sum())
```

```python
reco["rank_raw"] = reco["aggregate_rating"].rank(ascending=False, method="first")
reco["rank_weighted"] = reco["weighted_rating"].rank(ascending=False, method="first")

moved = reco.assign(shift=reco["rank_raw"] - reco["rank_weighted"])
moved[moved["votes"] < 20].sort_values("shift")[cols + ["rank_raw", "rank_weighted"]].head(8)
```

```python
city_n = reco.groupby("city")["restaurant_id"].transform("count")
city_mean = reco.groupby("city")["aggregate_rating"].transform("mean")
country_mean = reco.groupby("country")["aggregate_rating"].transform("mean")

prior = np.where(city_n >= 30, city_mean, country_mean)

reco["weighted_rating"] = ((v / (v + m)) * R + (m / (v + m)) * prior).round(3)

reco.sort_values("weighted_rating", ascending=False)[cols].head(10)
```

```python
def recommend(country, city, price_range=None, cuisines=None,
              table_booking=False, online_delivery=False, top_n=3):
    pool = reco[(reco["country"] == country) & (reco["city"] == city)].copy()
    if table_booking:
        pool = pool[pool["has_table_booking"] == 1]
    if online_delivery:
        pool = pool[pool["has_online_delivery"] == 1]
    if pool.empty:
        return pool, "No rated restaurants match this city and these service needs."

    wanted = set(cuisines or [])
    pool["cuisine_match"] = pool["cuisine_list"].apply(
        lambda xs: len(wanted & set(xs)) / len(wanted) if wanted else 0.0)
    pool["score"] = pool["weighted_rating"] + 0.5 * pool["cuisine_match"]

    stages = [
        ("Exact match", 0, True),
        ("Relaxed: price range within ±1, cuisine kept", 1, True),
        ("Relaxed: price range within ±1, cuisine ignored", 1, False),
        ("Relaxed: any price range, cuisine ignored", 4, False),
    ]
    for label, tol, need_cuisine in stages:
        sub = pool
        if price_range is not None:
            sub = sub[(sub["price_range"] - price_range).abs() <= tol]
        if need_cuisine and wanted:
            sub = sub[sub["cuisine_match"] > 0]
        if len(sub) >= top_n:
            break

    show = ["restaurant_name", "locality", "cuisines", "price_range",
            "average_cost_for_two", "currency", "aggregate_rating",
            "votes", "weighted_rating"]
    return sub.sort_values("score", ascending=False).head(top_n)[show], label
```

```python
%%writefile app.py
import pandas as pd
import streamlit as st

st.set_page_config(page_title="Restaurant Recommender", page_icon="🍽️")

@st.cache_data
def load():
    d = pd.read_csv("reco_app.csv")
    d["cuisine_list"] = d["cuisines"].str.split(",").apply(lambda xs: [x.strip() for x in xs])
    return d

reco = load()

def recommend(country, city, price_range=None, cuisines=None,
              table_booking=False, online_delivery=False, top_n=3):
    pool = reco[(reco["country"] == country) & (reco["city"] == city)].copy()
    if table_booking:
        pool = pool[pool["has_table_booking"] == 1]
    if online_delivery:
        pool = pool[pool["has_online_delivery"] == 1]
    if pool.empty:
        return pool, "No rated restaurants match this city and these service needs."

    wanted = set(cuisines or [])
    pool["cuisine_match"] = pool["cuisine_list"].apply(
        lambda xs: len(wanted & set(xs)) / len(wanted) if wanted else 0.0)
    pool["score"] = pool["weighted_rating"] + 0.5 * pool["cuisine_match"]

    stages = [
        ("Exact price range, cuisine required", 0, True),
        ("Relaxed: price range within ±1, cuisine required", 1, True),
        ("Relaxed: price range within ±1, cuisine ignored", 1, False),
        ("Relaxed: any price range, cuisine ignored", 4, False),
    ]
    for label, tol, need_cuisine in stages:
        sub = pool
        if price_range is not None:
            sub = sub[(sub["price_range"] - price_range).abs() <= tol]
        if need_cuisine and wanted:
            sub = sub[sub["cuisine_match"] > 0]
        sub = sub.sort_values("score", ascending=False).drop_duplicates("restaurant_name")
        if len(sub) >= top_n:
            break
    return sub.head(top_n), label

st.title("🍽️ Restaurant Recommender")
st.caption("Built on a Zomato dataset. It is an older snapshot, about 90% from the Delhi region, "
           "so some restaurants may have closed.")

countries = sorted(reco["country"].unique())
country = st.sidebar.selectbox("Country", countries, index=countries.index("India"))

cities = sorted(reco.loc[reco["country"] == country, "city"].unique())
city = st.sidebar.selectbox("City", cities)

in_city = reco[(reco["country"] == country) & (reco["city"] == city)]
if len(in_city) < 30:
    st.sidebar.warning(f"Only {len(in_city)} rated restaurants here, so choices are limited.")

any_price = st.sidebar.checkbox("Any price range")
price = None if any_price else st.sidebar.select_slider(
    "Price range (1 = cheapest, 4 = priciest)", options=[1, 2, 3, 4], value=2)

cuisine_options = sorted(in_city["cuisine_list"].explode().dropna().unique())
cuisines = st.sidebar.multiselect("Cuisines (optional)", cuisine_options)
table = st.sidebar.checkbox("Needs table booking")
delivery = st.sidebar.checkbox("Needs online delivery")

if st.sidebar.button("Recommend", type="primary"):
    res, note = recommend(country, city, price, cuisines, table, delivery)
    if res.empty:
        st.warning(note)
    else:
        (st.info if note.startswith("Relaxed") else st.success)(note)
        for _, r in res.iterrows():
            st.subheader(r["restaurant_name"])
            st.write(f"📍 {r['locality']}  ·  🍴 {r['cuisines']}")
            cost = r["average_cost_for_two"]
            cost_txt = "n/a" if pd.isna(cost) else f"{int(cost)} {r['currency']}"
            c1, c2, c3, c4 = st.columns(4)
            c1.metric("Rating", r["aggregate_rating"])
            c2.metric("Votes", int(r["votes"]))
            c3.metric("Cost for two", cost_txt)
            c4.metric("Price range", int(r["price_range"]))
            st.caption(f"Table booking: {'Yes' if r['has_table_booking'] == 1 else 'No'}  ·  "
                       f"Online delivery: {'Yes' if r['has_online_delivery'] == 1 else 'No'}")
            st.divider()
else:
    st.write("Choose your preferences in the sidebar, then press **Recommend**.")
```

```python
export_cols = ["restaurant_id", "restaurant_name", "locality", "city", "country",
               "cuisines", "price_range", "average_cost_for_two", "currency",
               "aggregate_rating", "votes", "weighted_rating",
               "has_table_booking", "has_online_delivery", "latitude", "longitude"]

reco[export_cols].to_json("restaurants.json", orient="records", force_ascii=False)
print(len(reco))
```
